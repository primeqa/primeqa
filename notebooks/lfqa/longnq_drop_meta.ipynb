```python
import json
import gzip

def load_json_from_file(gt_file_patterns, count=-1):
    data = []
    if gt_file_patterns.endswith('gz'):
        f = gzip.open(gt_file_patterns, 'rt', encoding='utf-8')
    else:
        f = open(gt_file_patterns, 'rt', encoding='utf-8')
    lines = f.readlines()
    for line in lines[0:count]:
        data.append(json.loads(line))
    return data

def write_jsonl(filename: str, content: list[dict], encoding="utf-8"):
    with open(filename, mode="w", encoding=encoding) as fp:
        for entry in content:
            fp.write(f"{json.dumps(entry)}\n")

json_data = load_json_from_file("/dccstor/srosent2/generative/appen/final/longNQ_hf/train/longNQ_train_answerable.jsonl")
json_data[0]
```
