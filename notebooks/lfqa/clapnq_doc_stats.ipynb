```python
import json

def read_json(input_file):
    with open(input_file, mode="r", encoding='utf-8') as fp:
        data = [json.loads(line.rstrip("\n").strip()) for line in fp]
    return data

nq_dev_ids = set()

test_data = read_json("/dccstor/srosent2/generative/appen/final/longNQ/test/longNQ_test_answerable_wdev.jsonl")

for example in test_data:
    if example['is_dev']:
        nq_dev_ids.add(example['id'])

print(len(nq_dev_ids))
```

```python
# load the training data and see how many in train/ dev / test have overlapping passages
import glob

data_files = glob.glob("/dccstor/srosent2/generative/appen/final/original_tydi/test/*.jsonl") # _answerable*
data_files.extend(glob.glob("/dccstor/srosent2/generative/appen/final/original_tydi/dev/*.jsonl"))
data_files.extend(glob.glob("/dccstor/srosent2/generative/appen/final/original_tydi/train/*.jsonl"))

data_by_title = {}
split_count = 0


for file_name in data_files:
    data = read_json(file_name)
    # if 'train' in file_name:
    #     split = 'train'
    # elif 'dev' in file_name:
    #     split = 'dev'
    # else:
    #     split = 'test'
    
    for doc in data:
        split = 'train'
 
        if doc['example_id'] in nq_dev_ids:
           split = 'dev'
           split_count += 1

        title = doc['document_title']

        if title not in data_by_title:
            data_by_title[title] = {} 
        if split not in data_by_title[title]:
            data_by_title[title][split] = 1
        else:
            data_by_title[title][split] += 1

print(split_count)
```

```python
duplicates_across_splits = 0
train_test = 0
train_dev = 0
dev_test = 0

for title in data_by_title:
    if len(data_by_title[title]) > 1:
        if 'train' in data_by_title[title] and 'test' in data_by_title[title]:
            train_test += 1
        if 'train' in data_by_title[title] and 'dev' in data_by_title[title]:
            train_dev += 1
        if 'dev' in data_by_title[title] and 'test' in data_by_title[title]:
            dev_test += 1
        # print(data_by_title[title])
        duplicates_across_splits += 1
print(f"duplicates: {duplicates_across_splits}")
print(f"train test {train_test}")
print(f"train dev {train_dev}")
print(f"dev test {dev_test}")
```
